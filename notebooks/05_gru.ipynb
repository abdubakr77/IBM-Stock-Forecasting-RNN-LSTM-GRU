# GRU Model Training

```python
# path setup so we can import from src
import os
import sys
sys.path.append(os.path.abspath('..'))

from src.utils import load_model_data
from src.model_gru import build_gru,build_attention_with_gru
from src.train import train_model
```

## Config
All the hyperparameters for this model in one place, tuned separately for each architecture.

Note: the model variable is still named `lstm_model` here, kept as is from the original code, it holds the GRU model.

```python
num_features = 8
timesteps = 90
target_index = 3
data_processed_dir = '../data/processed'
save_model_dir = '../models'

lr, weight_decay = 5e-4 , 5e-4
patience = 10
dropout = 0
batch_size = 256
epochs = 100
```

## Load the preprocessed data
Same train/valid/test splits saved from the preprocessing notebook, already scaled and windowed into sequences.

```python
X_train, y_train, X_valid, y_valid, X_test, y_test = load_model_data(os.path.join(data_processed_dir,'all_model_data.pkl'))
```

## Build the GRU model

```python
gru_model, gru_callbacks = build_gru(timesteps,
                                     num_features,
                                     dropout=dropout,
                                     lr=lr,
                                     weight_decay=weight_decay,
                                     patience=patience)
```

## Train the model
Saves the trained model and returns its training history for later comparison.

```python
gru_model, gru_history = train_model(gru_model,
                                    [X_train, y_train],
                                    [X_valid, y_valid],
                                    epochs=epochs,
                                    batch_size=batch_size,
                                    callbacks=gru_callbacks,
                                    model_name= 'GRU',
                                    save_dir=save_model_dir
                                    )
```
