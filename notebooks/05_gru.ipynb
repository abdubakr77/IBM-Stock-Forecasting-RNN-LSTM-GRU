```python
# path setup so we can import from src
import os
import sys
sys.path.append(os.path.abspath('..'))

from src.utils import load_model_data
from src.model_gru import build_gru
from src.train import train_model
```

```python
num_features = 8
timesteps = 90
lr, weight_decay = 5e-4 , 5e-4
patience = 10
dropout = 0

batch_size = 256
epochs = 100

target_index = 3

data_processed_dir = '../data/processed'
save_model_dir = '../models'
```

```python
X_train, y_train, X_valid, y_valid, X_test, y_test = load_model_data(os.path.join(data_processed_dir,'all_model_data.pkl'))
```

```python
lstm_model, callbacks = build_gru(timesteps,num_features,
                                    dropout=dropout,
                                    lr=lr,
                                    weight_decay=weight_decay,
                                    patience=patience)
```

```python
models_dict, history_dict = train_model([lstm_model],
                                        [X_train, y_train],
                                        [X_valid, y_valid],
                                        epochs=epochs,
                                        batch_size=batch_size,
                                        callbacks=callbacks,
                                        save_dir=save_model_dir
                                        )
```
